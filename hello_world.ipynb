```python
%%writefile hello_agent.py
from bedrock_agentcore import BedrockAgentCoreApp
from strands import Agent

app = BedrockAgentCoreApp()
agent = Agent(system_prompt="You are a friendly assistant. Keep answers short.")

@app.entrypoint
def invoke(payload):
    prompt = payload.get("prompt", "Hello!")
    result = agent(prompt)
    return {"result": result.message}

if __name__ == "__main__":
    app.run()
```

```python
import boto3
from bedrock_agentcore_starter_toolkit.notebook import Runtime

region = boto3.Session().region_name or "us-east-1"
runtime = Runtime()

runtime.configure(
    entrypoint="hello_agent.py",
    agent_name="hello_strands_agent",
    requirements_file="requirements.txt",
    region=region,
    auto_create_execution_role=True,
    auto_create_ecr=True,
)
```

```python
launch_result = runtime.launch()
print(launch_result.agent_arn)
```
